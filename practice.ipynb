```python
# Check if a number is a Prime
def is_prime(n):
    if n<2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n%i == 0:
            return False
    return True

print(is_prime(42))
print(is_prime(37))
```

```python
#Print the first N prime numbers
def first_n_primes(n):
    primes, num = [], 2
    while len(primes) < n:
        if num < 2:
            num += 1
            continue
        # Check if 'num' is prime
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
        num += 1
    return primes

print(first_n_primes(15))
```

```python
#Generate Fibonacci sequence upto N terms
def fibonacci(n):
    fib = [0,1]
    for i in range(2, n):
        fib.append(fib[-1] + fib[-2])
    return fib[:n]


print(fibonacci(20))
```

```python
#Check if a number is a palindrome
def is_palindrome_number(n):
    return str(n) == str(n)[::-1]

palindromes = []
for n in range(1,1000):
    if is_palindrome_number(n) == True:
        palindromes.append(n)
print(len(palindromes))
```

```python
#5 Find the Factorial of a number
def factorial(n):
    result = 1
    for i in range(2, n+1):
        result *= i
    return result

print(factorial(5))
```

```python
#6 Check if a number is an Armstrong number
def is_armstrong(n):
    digits = [int(d) for d in str(n)]
    power = len(digits)
    return n == sum(d ** power for d in digits)

arm_num = []
for n in range(1, 100000):
    if is_armstrong(n) == True:
        arm_num.append(n)

print(arm_num)
```

```python
#7 Find GCD of two numbers
def gcd(a,b):
    while b:
        a, b = b, a%b
    return a

print(gcd(18, 48))
print(gcd(48, 18))
```

```python
#8 Find LCM of 2 numbers
def lcm(a, b):
    return abs(a*b)//gcd(a,b)

print(lcm(6, 14))
```

```python
def lcm(a, b):
    maximum = max(a, b)
    while True:
        if maximum % a == 0 and maximum % b == 0:
            return maximum
        maximum += 1

print(lcm(6, 14))
```

```python
#9 Sum of digits of a number
def sum_of_digits(n):
    return sum(int(d) for d in str(n))

print(sum_of_digits(576))
```

```python
#12 Count Vowels in a string
def count_vowels(s):
    vowels = "aeiouAEIOU"
    return sum(1 for c in s if c in vowels)

p = "How many vowels are in this line"
count_vowels(p)
```

```python
#13 Most frequent character in a string

from collections import Counter
import string

def most_frequent_char(s):
    s = ''.join(c for c in s if c in string.ascii_letters)
    count = Counter(s)
    return max(count, key=count.get) if count else None

p = "How many vowels are in this line"
most_frequent_char(p)
```

```python
#14 Remove duplicate characters from a string

def remove_duplicates(s):
    seen = set()
    result = []
    for char in s:
        if char not in seen:
            seen.add(char)
            result.append(char)
    return ''.join(result)

remove_duplicates('programming')
```

```python
#17 Sort a list without using sort()

def selection_sort(arr):
    for i in range(len(arr)):
        min_idx = i
        for j in range(i+1, len(arr)):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

r = ['d', 'u', 'q', 'r', 'i', 'b']
p = [4, 43, 1, 65, 34]
print(selection_sort(r))
selection_sort(p)
```

```python
#18 Remove duplicates from a list

def remove_duplicates_list(arr):
    return list(set(arr))

remove_duplicates_list([1, 3, 4, 4, 5, 6, 6])
```

```python
#19 Find the second largest number in a list

def second_largest(arr):
    first = second = float('-inf')
    for num in arr:
        if num > first:
            second, first = first, num
        elif first > num > second:
            second = num
    return second if second != float('-inf') else None

p = [4, 43, 1, 65, 34]
second_largest(p)
```

```python
#20 Count frequency of elements in a list

def frequency_count(arr):
    from collections import Counter
    return dict(Counter(arr))

r = [1, 2, 2, 3, 3, 3]
frequency_count(r)
```
